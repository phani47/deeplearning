```python
import tensorflow as tf
from tensorflow import keras
import matplotlib.pyplot as plt
%matplotlib inline
import numpy as np
```

```python
(X_train,y_train) , (X_test,ytest) = keras.datasets.mnist.load_data()
```

```python
X_train = X_train / 255
X_test = X_test / 255
```

```python
X_train_flattened = X_train.reshape(len(X_train),28*28)
X_test_flattened = X_test.reshape(len(X_test),28*28)
```

```python
X_train_flattened[0]
```

```python
model = keras.Sequential([
    keras.layers.Dense(10, input_shape=(784,), activation='sigmoid')
])
```

```python
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

model.fit(
    X_train_flattened,
    y_train,
    epochs=20
)
```

```python
model.evaluate(X_test_flattened,ytest)
```

```python
plt.matshow(X_test[0])
```

```python
y_pred = model.predict(X_test_flattened)
```

```python
y_pred[0]
```

```python
np.argmax(y_pred[0])
```

```python
y_pred_labels = [int(np.argmax(i)) for i in y_pred]
y_pred_labels[:5]
```

```python
ytest[:5]
```

```python
cm = tf.math.confusion_matrix(labels=ytest,predictions=y_pred_labels)
cm
```

```python
import seaborn as sn
plt.figure(figsize=(10,7))
sn.heatmap(cm,annot=True,fmt='d')
plt.xlabel('Predicted')
plt.ylabel('Truth')
```

```python
model = keras.Sequential([
    keras.layers.Dense(100, input_shape=(784,), activation='relu'),
    keras.layers.Dense(10, activation='sigmoid')
])

model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

model.fit(
    X_train_flattened,
    y_train,
    epochs=20
)
```

```python
model.evaluate(X_test_flattened,ytest)
```

```python
import seaborn as sn
plt.figure(figsize=(10,7))
sn.heatmap(cm,annot=True,fmt='d')
plt.xlabel('Predicted')
plt.ylabel('Truth')
```

```python
model = keras.Sequential([
    keras.layers.Flatten(input_shape = (28,28)),
    keras.layers.Dense(100, activation='relu'),
    keras.layers.Dense(10, activation='sigmoid')
])

model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

model.fit(
    X_train,
    y_train,
    epochs=20
)
```
